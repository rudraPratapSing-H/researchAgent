```python


from pydantic import model_validator, computed_field, BaseModel, Field


class User(BaseModel):
    id: int 
    age: int 
    name: str = Field( min_length=1, max_length=100) # Name must be at least 1 character long 
    email: str 
    assets: list[str] = None # Default to an empty list if not provided
    @model_validator(mode="after")
    def validate_user(self):
        if self.age < 18:
            raise ValueError("User must be at least 18 years old.")
        if not self.email.endswith("@gmail.com"):
            raise ValueError("Email must be a Gmail address.")
        return self

    @computed_field
    @property
    def category(self) -> str:
        if self.age < 30:
            return "Young Adult"
        elif self.age < 60:
            return "Adult"
        else:
            return "Senior"




def create_user(user: User):
    print(f"User created: {user.name}, Age: {user.age}, Email: {user.email}, Category: {user.category}, and Assets: {user.assets}")

user_data = {
    "id": 1, 
    "age": 25,
    "name": "John Doe",
    "email": "john@gmail.com",
    "assets": ["car", "house"]
   }

user = User(**user_data) #** to unpack the dictionary into keyword arguments

create_user(user)
```
